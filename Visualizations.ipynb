```
All_candidates_final ="/FinalProject/All_Candidates_Cleaned"
All_cand_df = (spark
      .read
      .format("parquet")
      .option("path", All_candidates_final)
      .load())
```

```
Cand = All_cand_df.filter(All_cand_df.CAND_NAME.contains('TRUMP') | All_cand_df.CAND_NAME.contains('BIDEN'))
```

```
Cand.write.mode("overwrite").saveAsTable("Cand_tbl")
```
